```python
# Load the trained model pipeline
loaded_pipeline = joblib.load('study_materials/LearnML/regression_project/model/house_price_ridge_pipeline.pkl')
```

```python
# Read test.csv
df_test = pd.read_csv('study_materials/LearnML/regression_project/house-prices-advanced-regression-techniques/test.csv')
```

```python
# Remove Id's of the records and save it for later
ids = df_test['Id']
df_test = df_test.drop(columns='Id')
```

```python
# Make predictions
predictions = loaded_pipeline.predict(df_test)
preds = np.expm1(predictions)
```

```python
# Convert to dataframe for convenience
y_pred = pd.Series(preds)
pred_df = pd.concat([ids, y_pred], axis=1)
pred_df.rename(columns={0:'SalePrice'}, inplace=True)
pred_df.head()
```
